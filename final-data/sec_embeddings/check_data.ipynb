```python
import pickle
import numpy as np

with open("embeddings.pkl", "rb") as f:
    embeddings = pickle.load(f)

print("🔢 Số vector:", embeddings.shape[0])
print("📏 Chiều của mỗi vector:", embeddings.shape[1])
print("📚 Một vài embedding đầu tiên:")
print(embeddings[:1])  # hiển thị 2 embedding đầu tiên
```

```python
import faiss

index = faiss.read_index("faiss_index.bin")

# Số vector trong FAISS
print("🔢 Số vector trong index:", index.ntotal)

# Chiều của mỗi vector
print("📏 Chiều vector:", index.d)

# Kiểu FAISS index
print("⚙️ Loại index:", type(index))

# Lấy vector thứ 0
vec = index.reconstruct(0)
print("📚 Vector đầu tiên:", vec[:])  # hiển thị 10 giá trị đầu tiên
```
